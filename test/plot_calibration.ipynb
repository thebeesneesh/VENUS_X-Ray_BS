# BATMAN Current to M/Q Calibration

Calibration points plotted from the supplied arrays.

```python
import numpy as np
import matplotlib.pyplot as plt

calibration_current = np.array([36.655, 52.018, 56.322, 60.168, 68.882, 73.568, 84.77, 103.293])
calibration_mq = np.array([1.0, 2.0, 2.29, 2.67, 3.20, 4.0, 5.33, 8.0])

plt.figure(figsize=(8, 5))
plt.plot(calibration_current, calibration_mq, color='steelblue', linewidth=1.5, alpha=0.8)
plt.scatter(calibration_current, calibration_mq, color='darkorange', edgecolors='black', s=65, zorder=2)

for current, mq in zip(calibration_current, calibration_mq):
    plt.annotate(f'{mq:g}', (current, mq), textcoords='offset points', xytext=(5, 5))

plt.xlabel('BATMAN Current')
plt.ylabel('M/Q')
plt.title('BATMAN Current to M/Q Calibration')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
